## Definition
- this is a k-dimensional case of the Beta-distributed R.V.
- instead of 2 normalised gamma-distributed variables, a dirichlet random vector is a collection of k normalised gamma-distributed R.V.s
- $\forall \,,\, i \, \epsilon \, [1,2,...K] \,\,,\,\,  X_i \sim Gamma(\alpha_i, 1) \rightarrow Z_i = \dfrac{X_i}{\sum\limits_{i=1}^K X_i} \rightarrow Dirichlet(\alpha_1, \alpha_2 \cdots \alpha_K) = [Z_1, Z_2, \cdots Z_K]$
- the normalised random variables $Z_i$ are called **probability parameters** of the dirichlet distribution.

## P.D.F. Derivation
1. transformations: $T = \sum\limits_{i=1}^K X_i \,,\,\, Z_i = \dfrac{X_i}{T}$
    1. to keep the no. of variables to K, use this transformation as well: $X_K = T - \sum\limits_{i=1}^{K-1} X_i = T - \sum\limits_{i=1}^{K-1} TZ_i = T(1 - \sum\limits_{i=1}^{K-1}Z_i)$
    2. therefore, $X_i = T.Z_i \rightarrow \dfrac{\partial X_i}{\partial Z_j} = T.\delta_{ij} \,,\, \dfrac{\partial X_i}{\partial T} = Z_i \,, \, \forall \, i \, \epsilon \, [1,2,\cdots K-1]$\
    $\dfrac{\partial X_K}{\partial T} = 1 - \sum\limits_{i=1}^{K-1}Z_i \quad ,\quad \dfrac{\partial X_K}{\partial Z_i} = -T \quad \forall \, i \, \epsilon \, [1,2,\cdots K-1]$
2. Jacobian $J = \begin{bmatrix}\dfrac{\partial X_1}{\partial Z_1} & \dfrac{\partial X_1}{\partial Z_2} & \cdots & \dfrac{\partial X_1}{\partial Z_{K-1}} & \dfrac{\partial X_1}{\partial T} \\  \dfrac{\partial X_2}{\partial Z_1} & \dfrac{\partial X_2}{\partial Z_2} & \cdots & \dfrac{\partial X_2}{\partial Z_{K-1}} & \dfrac{\partial X_2}{\partial T} \\ & & \vdots &  \\ \dfrac{\partial X_K}{\partial Z_1} & \dfrac{\partial X_K}{\partial Z_2} & \cdots & \dfrac{\partial X_K}{\partial Z_{K-1}} & \dfrac{\partial X_K}{\partial T} \end{bmatrix} = \begin{bmatrix}T & 0 & \cdots & 0 & Z_1 \\  0 & T & \cdots & 0 & Z_2 \\ & & \vdots &  \\ -T & -T & \cdots & -T & 1 - \sum\limits_{i=1}^{K-1}Z_i \end{bmatrix}$
3. lets take a simple example, of $K = 4$ to know the pattern created
    1. for K = 4, $J = \begin{bmatrix} T & 0 & 0 & Z_1 \\ 0 & T & 0 & Z_2 \\ 0 & 0 & T& Z_3 \\ -T &-T &-T & 1-(Z_1 + Z_2 + Z_3)  \end{bmatrix}$
    2. determinant $|J| = (-1)^{1+1}T \begin{vmatrix} T & 0 & Z_2 \\ 0 & T & Z_3 \\ -T & -T & (1-Z_1 -Z_2 - Z_3) \end{vmatrix} + (-1)^{1+4}Z_1 \begin{vmatrix} 0 & T& 0 \\ 0 & 0 & T \\ -T & -T & -T \end{vmatrix} = T^3 - T^3\sum\limits_{i=1}^3 Z_i+ (-1)^{1+4}Z_1.(-1)^{1+2}.T.\begin{vmatrix}0 & T \\ -T & -T \\ \end{vmatrix} +  T.(-1)^{1+3}Z_2 \begin{vmatrix}0 & T \\ -T & -T \\ \end{vmatrix} + (-1)^{1+4}Z_1 (-1)^{1+2} T \begin{vmatrix}0 & T \\ -T & -T \\ \end{vmatrix} = T^3 - T^3\sum\limits_{i=1}^3 Z_i + (-1)^{1+4}Z_1.(-1)^{1+2}.T.(-1)^{1+2}.(-T^2) +  T.(-1)^{1+3}Z_2.(-1)^{1+2}.(-T^2) = T^{K-1} - T^{K-1} \sum\limits_{i=1}^{K-1} Z_i + \sum\limits_{i=1}^{K-1} \left[T^{K-1}Z_i . (-1)^{1 + K+1-i}.\left((-1)^{1+2}\right)^{K-1-i}\right]$
    3. $3^{rd}$ term's inner summation term: $T^{K-1}Z_i . (-1)^{1 + K+1-i}.\left((-1)^{1+2}\right)^{K-1-i}.(-1) = T^{K-1}Z_i.(-1)^{\large K+3-i + 3K-3-3i} = T^{K-1}Z_i.(-1)^{\large 4K-4i} = T^{K-1}Z_i$
    4. Therefore, for K = 4, $|J| = T^{K-1} - T^{K-1} \sum\limits_{i=1}^{K-1} Z_i + \sum\limits_{i=1}^{K-1} T^{K-1}Z_i = T^{K-1}$
4. $|J| = T^{K-1}$
5. $f_{Z_1, Z_2, \cdots \, Z_{K-1}, T} = f_{X_1, X_2, \cdots X_K}.|J| = T^{K-1}.\prod\limits_{i=1}^K \dfrac{X_i^{\alpha_i - 1}e^{-X_i}}{\Gamma(\alpha_i)} = T^{K-1}.e^{-\sum\limits_{i=1}^K X_i} \prod\limits_{i=1}^K \dfrac{X_i^{\alpha_i - 1}}{\Gamma(\alpha_i)} = T^{K-1}e^{-T} \prod\limits_{i=1}^K \dfrac{X_i^{\alpha_i - 1}}{\Gamma(\alpha_i)} = T^{K-1}e^{-T} \left(\prod\limits_{i=1}^{K-1} \dfrac{Z_i^{\alpha_i - 1}T^{\alpha_i - 1}}{\Gamma(\alpha_i)}\right)\dfrac{T^{\alpha_K - 1}(1 - \sum\limits_{i=1}^{K-1} Z_i)^{\alpha_K - 1}}{\Gamma(\alpha_K)} = T^{K-1}e^{-T} T^{K - 1 + \left(\sum\limits_{i=1}^K\alpha_i \right) - K} \prod\limits_{i=1}^{K-1} \dfrac{Z_i^{\alpha_i - 1}T^{\alpha_i - 1}}{\Gamma(\alpha_i)} = T^{\left(\sum\limits_{i=1}^K\alpha_i \right) - 1}e^{-T} \prod\limits_{i=1}^{K} \dfrac{Z_i^{\alpha_i - 1}}{\Gamma(\alpha_i)}$
6. marginalizing over $T$, $f_{Z_1, Z_2, \cdots Z_K} = \int\limits_{0}^{\infty} T^{\left(\sum\limits_{i=1}^K\alpha_i \right) - 1}e^{-T} \prod\limits_{i=1}^{K} \dfrac{Z_i^{\alpha_i - 1}}{\Gamma(\alpha_i)}dT = \prod\limits_{i=1}^{K} \dfrac{Z_i^{\alpha_i - 1}}{\Gamma(\alpha_i)} \int\limits_{0}^{\infty} T^{\left(\sum\limits_{i=1}^K\alpha_i \right) - 1}e^{-T} dT = \dfrac{\Gamma(\sum\limits_{i=1}^K \alpha_i)}{\prod\limits_{i=1}^K \Gamma(\alpha_i)} . Z_1^{\alpha_1 - 1}.Z_2^{\alpha_2 - 1}.\cdots.Z_K^{\alpha_K - 1}$
7. $f_{\left[Z_1, Z_2, \cdots Z_K\right]} = \dfrac{\Gamma\left(\sum\limits_{i=1}^K \alpha_i \right)}{\prod\limits_{i=1}^K \Gamma(\alpha_i)} . Z_1^{\alpha_1 - 1}.Z_2^{\alpha_2 - 1}.\cdots.Z_K^{\alpha_K - 1}$

# Exploration via code

```python
import numpy as np
from scipy.stats import dirichlet, expon, uniform
from scipy.special import softmax

import matplotlib.pyplot as plt
import seaborn as sb
```

```python
np.random.seed(42)
K = 3
x = uniform.rvs(0, 1, size = (100))
y = uniform.rvs(0.5, 1.5, size = (100))
z = uniform.rvs(1, 3, size = (100))

# the random variable that the pdf of dirichlet
# distribution takes as an input isn't a scalar
# but rather a vector
x_grid, y_grid, z_grid = np.meshgrid(x, y, z)

# np.hstack simply stacks these 1d-arrays beside each other, thus generating
# a (3M,) shaped 1d-array
# alternative to np.column_stack: np.stack.T
x_logit = np.stack([x_grid.flatten(), y_grid.flatten(), z_grid.flatten()])

# a random-variable vector should have all components summing to 1, 
# which is ensured by this transformation
x_softmaxed = softmax(x_logit.T, axis=1).T
print(x_softmaxed.shape)

sb.kdeplot(x_softmaxed[0, :], label='theta_1')
sb.kdeplot(x_softmaxed[1, :], label='theta_2')
sb.kdeplot(x_softmaxed[2, :], label='theta_3')

plt.legend()
plt.show()
```

The above kdeplot does look a lot like the P.D.F. of a dirichlet distribution shown below $\downarrow$

```python
np.random.seed(42)

alpha = expon.rvs(size = 3)
dirichlet_sample = dirichlet.rvs(alpha = alpha, size = 1000)
print(dirichlet_sample[:10, :])

sb.kdeplot(dirichlet_sample[:, 0], label=f'alpha_1 = {alpha[0]}')
sb.kdeplot(dirichlet_sample[:, 1], label=f'alpha_2 = {alpha[1]}')
sb.kdeplot(dirichlet_sample[:, 2], label=f'alpha_3 = {alpha[2]}')

plt.legend()
plt.show()
```
